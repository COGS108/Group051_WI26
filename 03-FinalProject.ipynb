## Abstract

In rhythm games, new players have the tendency to struggle with an appropriate difficulty to try out and are often left discouraged to pursue a rhythm game further. One instance in particular would be in the rhythm game Dance Dance Revolution (DDR), where overlapping labels can make it hard for a new player to identify their current skill level. Our project aims to investigate whether the performance on a DDR chart can reliably predict a player’s skill across charts of varying difficulty. Using gameplay statistics gathered from experienced DDR players sharing their scores, we analyzed how core skills used for DDR can potentially be generalized across DDR’s numeric ratings 1-20. We found the relevance of chart ratings pushing the extent of core skills as the difficulty goes up to accuracy metrics like EX scores (DDR’s accuracy based scoring system); applying statistical models to evaluate the strength of this relationship. Using our foundations of DDR’s experienced playerbase at UCSD’s annual animefest event to emphasize data science in practice - potentially using our work to get more students at UCSD into rhythm games.
